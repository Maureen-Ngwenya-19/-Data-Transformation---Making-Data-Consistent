```python
import pandas as pd
```

```python
ages = [18,22,25,30,35,42,47,52,61,68,75]

print(ages)
```

```python
#Define age groups(bins)

age_bins =[0,25,35,50,65,100]

age_labels =['Youth (0-25)','Young Adult (26-35)', 'Middle Age (36-50)','senior (51-65)' , 'Elderly (66+)']
```

```python
# categorize ages 
# use pd.cut to categorize each group in the ages list into bins defined by
 
age__categories = pd.cut(ages, bins=age_bins, labels=age_labels)
print("Age converted to categories :")

# Loop through each age and its corresponding category in the categorized result
for age,category in zip(ages,age__categories):
      # Print the age and its assigned category
    print(f"Age {age} : {category}")
```
